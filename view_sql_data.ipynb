```python
import sqlite3
import pandas as pd
from pathlib import Path

# 1. Define the relative path to the database
# Using pathlib ensures Windows/Mac/Linux path compatibility
db_path = Path("data/processed/erp_database.sqlite")

# 2. Create a connection to the SQLite database
conn = sqlite3.connect(db_path)

# Optional: List all tables in the database if you don't know the table name
tables_query = "SELECT name FROM sqlite_master WHERE type='table';"
tables_df = pd.read_sql_query(tables_query, conn)
print("Tables in database:")
print(tables_df)
```

```python

# 3. Read a specific table into a DataFrame
# Replace 'your_table_name' with the actual name of your table
table_name = "ai_reports" 
query = f"SELECT * FROM {table_name}"

# Load the data
df = pd.read_sql_query(query, conn)

# 4. Close the connection
conn.close()

# Display the first few rows
# print(f"\nData from {table_name}:")
# print(df.head())
```
